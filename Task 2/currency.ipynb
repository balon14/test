```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
import seaborn as sns
from pathlib import Path
warnings.simplefilter(action='ignore', category=FutureWarning)
```

```python
files = {
    "2027": "D:/_test/курс_валют.xlsx",
}

ex = {name: pd.read_excel(path) for name, path in files.items()}

df = ex["2027"]
```

```python
df
```

```python
df['Дата'] = pd.to_datetime(df['Дата'])
```

```python
# 3. Задаем параметры деноминации
date_denom = pd.to_datetime('2016-07-01') # Дата деноминации
coefficient = 10000                        # Коэффициент (во сколько раз уменьшился курс)

# 4. Создаем новый столбец с единым форматом
# Если дата меньше дня деноминации — делим на коэффициент, иначе — оставляем как есть
df['Курс_новый'] = df.apply(
    lambda row: row['1 доллар США (USD)'] / coefficient if row['Дата'] < date_denom else row['1 доллар США (USD)'], 
    axis=1
)

# Смотрим результат
print(df)
```

```python
df.to_excel('currency_fixed.xlsx', index=False)
```

```python
# 1. Делаем столбец с датами индексом таблицы (это необходимо для правильной работы resample)
df.set_index('Дата', inplace=True)

# 2. Группируем по месяцам ('ME' или 'M') и находим среднее значение курса
# Параметр 'ME' означает Month End (конец месяца) — он сгруппирует данные по месяцам
df_monthly = df[['Курс_новый']].resample('ME').mean()

# 3. Возвращаем дату обратно из индекса в обычный столбец, чтобы таблица выглядела красиво
df_monthly.reset_index(inplace=True)

# Округляем курс, например, до 4 знаков после запятой (по желанию)
df_monthly['Курс_новый'] = df_monthly['Курс_новый'].round(4)

# Смотрим результат в консоли
print(df_monthly)

# 4. Сохраняем сжатую таблицу в новый файл Excel
df_monthly.to_excel('currency_monthly_average.xlsx', index=False)
```
